## 🧱 Data Structure–Specific Algorithms

### 1️⃣ Arrays

#### Sorting
- **QuickSort**
  - Efficient average-case performance  
  - Time Complexity: `O(n log n)`
- **MergeSort**
  - Stable sorting algorithm  
  - Useful when relative order matters  
  - Time Complexity: `O(n log n)`
- **InsertionSort**

- **BubbleSort**

- **RadixSort**

- **BubbleSort**

#### Searching
- **Binary Search**
  - Works on sorted arrays  
  - Time Complexity: `O(log n)`

#### Two Pointers
- In-place array manipulation  
- Common use cases:
  - Removing duplicates
  - Pair-sum problems
  - Sorted array operations

#### Sliding Window
- Efficient solution for subarray problems
- Use cases:
  - Maximum/minimum in a window
  - Longest/shortest subarray satisfying constraints

## ⚙️ General Algorithms & Techniques

### 1️⃣ Recursion
- Define problems in terms of themselves
- Leads to elegant solutions

**Solve:**
- Factorial
- Tree traversals
- Depth-First Search

---

### 2️⃣ Dynamic Programming
- Break problems into overlapping subproblems
- Store results to avoid recomputation

**Solve:**
- Fibonacci sequence
- Knapsack problem
- Longest Common Subsequence (LCS)

---

### 3️⃣ Greedy Algorithms
- Make locally optimal choices
- Hope to reach global optimum

**Implement:**
- Kruskal’s Algorithm (Minimum Spanning Tree)

---

### 4️⃣ Backtracking
- Build solutions incrementally
- Explore all possibilities
- Abandon invalid paths early

**Solve:**
- Sudoku Solver
- N-Queens problem
- Generating permutations
